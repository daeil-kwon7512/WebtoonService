```python
genre_set = set()
for i in json_data:
    genre_set = genre_set | set(i.get("genre").split(','))
```

```python
genre_set = [i.lstrip(" ") for i in genre_set]
```

```python
raw_df = pd.DataFrame(genre_set).sort_values(by=0).reset_index(drop=True).drop_duplicates()
```

```python
for i in raw_df.values:
    print(i)
    break
```

```python
genre_ = {"NAVER":set(), "KAKAO":set()}
for genre_tag in raw_df.values:
    provider_name = collections.Counter({"NAVER":0, "KAKAO":0, "KAKAOPAGE":0})
    for i in json_data:
        if genre_tag in [j.strip() for j in i.get("genre").split(',')]:
            # print(i)
            provider_name.update([j.strip() for j in i.get("provider").split(',')])
            if provider_name["NAVER"] == 0:
                genre_["KAKAO"].add(genre_tag[0])
            elif provider_name["KAKAO"] == 0 and provider_name["KAKAOPAGE"] == 0:
                genre_["NAVER"].add(genre_tag[0])
```
